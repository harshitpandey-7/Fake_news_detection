### Problem 1: Word Count Ignoring Punctuation (Built-in Functions)

**Task:** Write a Python function that counts the number of words in a given string, ignoring all punctuation. Only consider sequences of alphanumeric characters as words. You should only use Python's built-in string methods (no regex or NLP libraries).

```python
text = "Hello, world! How are you doing today? Let's count these words (properly)."

# Solution: Strip punctuation using str.isalnum() and split into words
def count_words_no_punct(text):
    # Remove all non-alphanumeric characters (except spaces)
    cleaned = ''.join(char if char.isalnum() or char.isspace() else ' ' for char in text)
    # Split on whitespace and filter out empty strings
    words = cleaned.split()
    return len(words)

print("Word Count:", count_words_no_punct(text))
# Expected output for the above text: 10
```

**Explanation:** We iterate through each character. If it's alphanumeric or whitespace, we keep it; otherwise, we replace it with a space. Then `split()` handles multiple spaces and gives us clean words. "Let's" becomes "Let" and "s" (two words), and "(properly)" becomes "properly".

### Problem 2: Extracting URLs (Regex)

**Task:** Use regular expressions to find and extract all URLs (web addresses starting with `http://` or `https://`) from a given text.

```python
import re

text = "Visit Google at https://www.google.com or check out Wikipedia at http://en.wikipedia.org for more info. My site is example.com"

# Solution: Use regex pattern to match URLs starting with http:// or https://
pattern = r'https?://[^\s]+'
urls = re.findall(pattern, text)

print("Extracted URLs:", urls)
# Expected output: ['https://www.google.com', 'http://en.wikipedia.org']
```

**Explanation:** The regex `https?://[^\s]+` matches:
- `http` – literal text
- `s?` – optional 's' (matches both http and https)
- `://` – literal '://'
- `[^\s]+` – one or more non-whitespace characters (the rest of the URL)

Note: `example.com` is NOT matched because it doesn't start with `http://` or `https://`.

### Problem 4: Stemming a List of Words (NLTK PorterStemmer)

**Task:** Apply NLTK's `PorterStemmer` to a list of words and print the stemmed versions of each word.

```python
import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["running", "runs", "runner", "easily", "fairly", "universal", "universities"]

# Solution: Apply stemmer to each word and print original vs stemmed
for word in words:
    stemmed = stemmer.stem(word)
    print(f"{word} -> {stemmed}")
```

**Explanation:** The Porter Stemmer applies a set of rules to strip suffixes from words and reduce them to their stems. For example:
- "running" → "run", "runs" → "run", "runner" → "runner"
- "easily" → "easili", "fairly" → "fairli"

Note: Stemming can produce non-dictionary words (like "easili"). For more linguistically accurate reduction, use lemmatization (see Problem 5).

### Problem 8: Top N Most Common Words (Counter)

**Task:** Find and print the 5 most common words in the provided text. Ensure that words are lowercased and punctuation is removed before counting. Use `collections.Counter`.

```python
from collections import Counter
import re

text = "This is a sample text. This text is just a sample for demonstration purposes. Sample text is good."

# Solution: Clean text, count word frequencies, and get top 5
# Step 1: Convert to lowercase
text_lower = text.lower()

# Step 2: Remove punctuation using regex (keep only alphanumeric and spaces)
text_clean = re.sub(r'[^\w\s]', '', text_lower)

# Step 3: Split into words
words = text_clean.split()

# Step 4: Count word frequencies
word_counts = Counter(words)

# Step 5: Get the 5 most common words
top_5 = word_counts.most_common(5)

print("Top 5 Most Common Words:")
for word, count in top_5:
    print(f"  '{word}': {count} times")
```

**Explanation:** The pipeline is: lowercase → remove punctuation → split → count. `Counter.most_common(5)` returns the 5 words with the highest frequency as a list of `(word, count)` tuples.

### Problem 9: Custom Stop Word Removal

**Task:** Create a function that removes a predefined list of stop words from a given sentence. The function should return the cleaned sentence as a string.

```python
stop_words = ["is", "a", "the", "and", "with", "some"]
sentence = "This is a sample sentence with some stop words that need to be removed."

# Solution: Filter out stop words from the sentence
def remove_stop_words(sentence, stop_words):
    words = sentence.split()
    filtered_words = [word for word in words if word.lower() not in stop_words]
    return ' '.join(filtered_words)

cleaned_sentence = remove_stop_words(sentence, stop_words)
print("Original:", sentence)
print("Cleaned: ", cleaned_sentence)
# Expected output: "This sample sentence stop words that need to be removed."
```

**Explanation:** We split the sentence into words, then use a list comprehension to keep only words whose lowercase form is NOT in the stop words list. Finally, we join the remaining words back into a string. The comparison uses `.lower()` to handle case-insensitive matching.

### Problem 11: Text Cleaning - Remove Numbers and Extra Spaces (Regex)

**Task:** Write a function that uses regular expressions to perform two cleaning steps on a text:
1.  Remove all digits (0-9).
2.  Replace multiple spaces with a single space.

The function should return the cleaned text.

```python
import re

text = "This text has 123 numbers and   extra   spaces, also 456 digits."

# Solution: Use regex to remove digits and collapse extra spaces
def clean_text(text):
    # Step 1: Remove all digits
    text = re.sub(r'\d', '', text)
    # Step 2: Replace multiple spaces with a single space
    text = re.sub(r'\s+', ' ', text)
    # Step 3: Strip leading/trailing whitespace
    text = text.strip()
    return text

cleaned = clean_text(text)
print("Original:", text)
print("Cleaned: ", cleaned)
# Expected output: "This text has numbers and extra spaces, also digits."
```

**Explanation:**
- `re.sub(r'\d', '', text)` – removes all digits (0-9) by replacing them with empty string
- `re.sub(r'\s+', ' ', text)` – replaces one or more whitespace characters with a single space
- `.strip()` – removes any leading/trailing whitespace

This is a common text cleaning pipeline used before feeding text into NLP models.

### Problem 12: Average Sentence Length (NLTK)

**Task:** Calculate the average word count per sentence in a given paragraph using NLTK's `sent_tokenize` and `word_tokenize`. Print the average length.

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Uncomment the following line if 'punkt' is not already downloaded
# nltk.download('punkt')
# nltk.download('punkt_tab')

paragraph = "NLP is amazing. It helps machines understand human language. This is a very interesting field of study and has many applications."

# Solution: Tokenize into sentences, count words per sentence, compute average
sentences = sent_tokenize(paragraph)
total_words = 0

print("Sentence-wise word count:")
for i, sentence in enumerate(sentences, 1):
    words = word_tokenize(sentence)
    word_count = len(words)
    total_words += word_count
    print(f"  Sentence {i}: '{sentence}' -> {word_count} words")

average_length = total_words / len(sentences)
print(f"\nAverage Sentence Length: {average_length:.2f} words")
```

**Explanation:** We first split the paragraph into sentences using `sent_tokenize`, then count words in each sentence using `word_tokenize`. The average is computed as total words / number of sentences. This metric is useful for readability analysis (e.g., Flesch reading ease score).

### Problem 13: Enhanced Rule-based Sentiment Analysis

**Task:** Modify the existing `sentiment_analysis` function to include a 'Mixed Sentiment' category. If the text contains both positive and negative words (and neither count is zero), it should be classified as 'Mixed Sentiment'. Otherwise, use the existing 'Positive', 'Negative', or 'Neutral' classifications.

```python
positive_words = {"happy", "good", "great", "excellent", "love", "amazing"}
negative_words = {"sad", "bad", "terrible", "hate", "awful", "worst"}

def sentiment_analysis_enhanced(text):
    # Tokenize and lowercase the text
    words = text.lower().split()
    # Remove punctuation from each word for accurate matching
    cleaned_words = [''.join(char for char in word if char.isalnum()) for word in words]
    
    # Count positive and negative words
    pos_count = sum(1 for word in cleaned_words if word in positive_words)
    neg_count = sum(1 for word in cleaned_words if word in negative_words)
    
    # Classify sentiment
    if pos_count > 0 and neg_count > 0:
        return "Mixed Sentiment"
    elif pos_count > neg_count:
        return "Positive Sentiment"
    elif neg_count > pos_count:
        return "Negative Sentiment"
    else:
        return "Neutral Sentiment"

text1 = "I love this product, but the delivery was terrible."
text2 = "This is a great movie."
text3 = "The service was awful."
text4 = "It was okay."

print(f"Text 1 Sentiment: {sentiment_analysis_enhanced(text1)}")
print(f"Text 2 Sentiment: {sentiment_analysis_enhanced(text2)}")
print(f"Text 3 Sentiment: {sentiment_analysis_enhanced(text3)}")
print(f"Text 4 Sentiment: {sentiment_analysis_enhanced(text4)}")
```

**Explanation:** The enhanced function adds a "Mixed Sentiment" check **before** the positive/negative comparison:
1. If both `pos_count > 0` AND `neg_count > 0` → **Mixed Sentiment**
2. If only positive words found → **Positive Sentiment**
3. If only negative words found → **Negative Sentiment**
4. If no sentiment words found → **Neutral Sentiment**

We also strip punctuation from words before matching (e.g., "terrible." → "terrible") to improve accuracy.

Expected output:
- Text 1: Mixed Sentiment (contains "love" and "terrible")
- Text 2: Positive Sentiment (contains "great")
- Text 3: Negative Sentiment (contains "awful")
- Text 4: Neutral Sentiment (no sentiment words)


### Problem 14: Bag-of-Words Representation for a New Corpus

**Task:** Apply `CountVectorizer` to the new corpus provided. Print the feature names (vocabulary) and the Bag-of-Words (BoW) representation (as an array) for each document in the corpus.

```python
from sklearn.feature_extraction.text import CountVectorizer

new_corpus = [
    "Machine learning is a fascinating field",
    "Deep learning is a subset of machine learning and uses neural networks",
    "Neural networks are powerful tools for AI"
]

# Solution: Apply CountVectorizer to create Bag-of-Words representation
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(new_corpus)

# Print feature names (vocabulary)
feature_names = vectorizer.get_feature_names_out()
print("Vocabulary (Feature Names):")
print(feature_names)
print()

# Print BoW representation for each document
print("Bag-of-Words Representation:")
bow_array = bow_matrix.toarray()
for i, doc in enumerate(new_corpus):
    print(f"\nDocument {i+1}: '{doc}'")
    print(f"  BoW Vector: {bow_array[i]}")
```

**Explanation:** `CountVectorizer` converts text into a matrix of word counts:
1. `fit_transform()` learns the vocabulary from the corpus and transforms each document into a count vector
2. Each column represents a unique word from the vocabulary
3. Each row represents a document
4. Each cell contains the count of that word in that document

This is one of the simplest text-to-number representations used in machine learning.

### Problem 15: TF-IDF and Top 3 Keywords per Document

**Task:** Apply `TfidfVectorizer` to the provided corpus. Then, for each document, identify and print the top 3 most important keywords based on their TF-IDF scores.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

corpus_for_tfidf = [
    "The cat sat on the mat and purred happily",
    "A small dog ran in the park chasing a ball",
    "Cats and dogs are common pets and can be very playful"
]

# Solution: Apply TfidfVectorizer and find top 3 keywords per document
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus_for_tfidf)

# Get feature names
feature_names = tfidf_vectorizer.get_feature_names_out()

print("Top 3 Keywords per Document (by TF-IDF score):")
print("=" * 50)

for i, doc in enumerate(corpus_for_tfidf):
    print(f"\nDocument {i+1}: '{doc}'")
    
    # Get TF-IDF scores for this document
    tfidf_scores = tfidf_matrix[i].toarray().flatten()
    
    # Get indices of top 3 scores (sorted in descending order)
    top_3_indices = np.argsort(tfidf_scores)[::-1][:3]
    
    print("  Top 3 Keywords:")
    for rank, idx in enumerate(top_3_indices, 1):
        print(f"    {rank}. '{feature_names[idx]}' (TF-IDF: {tfidf_scores[idx]:.4f})")
```

**Explanation:** TF-IDF (Term Frequency-Inverse Document Frequency) measures how important a word is to a document within a corpus:
- **TF (Term Frequency):** How often a word appears in a document
- **IDF (Inverse Document Frequency):** How rare or common a word is across all documents
- Words that appear frequently in one document but rarely in others get higher TF-IDF scores

We use `np.argsort()[::-1][:3]` to get the indices of the 3 highest TF-IDF scores, then map them back to feature names. This technique is widely used for keyword extraction and document summarization.
