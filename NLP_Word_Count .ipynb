## Python Built-in Function - Count Words

###  Python Built-in Function - Count Words
This code snippet demonstrates a basic way to count words in a string using Python's built-in functionalities. It first splits the input text into a list of words using `text.split()` (which by default splits on whitespace) and then calculates the length of the resulting list.

### Use Case:
This method is useful for quick, simple word counts where punctuation and complex text structures are not a major concern. For instance, counting words in a sentence that is already clean or for a rough estimate of text length.

```python
text = "Natural Language Processing (NLP) is a fascinating field!"
words = text.split()
word_count = len(words)
print("Word Count:", word_count)
print(words)
```

## Regex - Handle Punctuation and Count Words

### Regular Expressions (Regex)
Regular expressions (regex) are sequences of characters that define a search pattern. Python's built-in `re` module provides operations for working with regular expressions.

### Use Case:
Regex is incredibly powerful for advanced text processing tasks, such as:
*   **Pattern Matching:** Searching for specific patterns within text (e.g., email addresses, phone numbers, dates).
*   **Data Validation:** Ensuring input data conforms to a specific format.
*   **Text Extraction:** Pulling out relevant information from unstructured text.
*   **Text Replacement:** Finding and replacing parts of a string based on a pattern.
*   **Complex Tokenization:** Splitting text into words or tokens while handling various punctuation, special characters, or custom delimiters, as demonstrated in the word counting example.

```python
import re
# regular expression

def count_words(text):
    words = re.findall(r'\b\w+\b', text)
    return len(words)

text = "Hello, world! Welcome->have to NLP."
print("Word Count:", count_words(text))
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk import word_tokenize

def count_words_nltk(text):
    words = word_tokenize(text)
    return len(words)

text = "Let's explore NLP with Python!"
print("Word Count:", count_words_nltk(text))
```

## Spacy

### Explanation: spaCy
spaCy is an industrial-strength NLP library. It processes text by creating a `Doc` object, which contains tokens. Each token has attributes like `is_punct` (to check if it's punctuation). The code iterates through the tokens in the `Doc` object and counts only those that are not punctuation.

### Use Case:
spaCy is highly efficient and provides rich linguistic annotations, making it suitable for production-grade NLP applications. It's used for tasks requiring more sophisticated text processing like named entity recognition, dependency parsing, and semantic similarity, where accurate tokenization and filtering of non-word tokens are crucial.

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def count_words_spacy(text):
    doc = nlp(text)
    words = [token.text for token in doc if not token.is_punct]
    return len(words)

text = "NLP is amazing, isn't it?"
print("Word Count:", count_words_spacy(text))
```

## Counter

### Explanation: Counter
The `collections.Counter` class is a specialized dictionary subclass for counting hashable objects. In this example, after splitting the text and converting words to lowercase, `Counter` efficiently counts the occurrences of each unique word, providing a frequency distribution.

### Use Case:
`Counter` is ideal for quickly determining the frequency of words or any other items in a list. It's widely used in text analysis for tasks like identifying the most common words in a document (e.g., for keyword extraction, topic modeling), or for basic statistical analysis of text data.

```python
from collections import Counter
def word_frequency(text):
    words = text.lower().split()
    return Counter(words)

text = "NLP is fun. NLP is powerful. NLP is the future."
print("Word Frequencies:", word_frequency(text))
```

## NLP Task 1 - Short vs Long sentences

###  NLP Task 1 - Short vs Long sentences
This task demonstrates how to categorize sentences based on their word count. It uses NLTK's `sent_tokenize` to split the text into individual sentences and then `word_tokenize` to count words within each sentence. Sentences with less than 5 words are classified as 'short', and others as 'long'.

### Use Case:
This type of analysis can be useful in content analysis, readability assessments, or summarization tasks. For example, identifying overly long sentences might indicate areas for simplification in educational material, while a high proportion of short sentences could suggest a direct or fragmented writing style.

```python
from nltk.tokenize import sent_tokenize

def categorize_sentences(text):
    sentences = sent_tokenize(text)
    categorized = {"short": [], "long": []}

    for sentence in sentences:
        word_count = len(word_tokenize(sentence))
        if word_count < 5:
            categorized["short"].append(sentence)
        else:
            categorized["long"].append(sentence)

    return categorized

text = "Hello! NLP is fun. Let's learn NLP together. It helps in many fields."
categories = categorize_sentences(text)
print("Short Sentences:", categories["short"])
print("Long Sentences:", categories["long"])
```

## NLP Task 2 - Rule based Sentiment Analysis

###  - Rule-based Sentiment Analysis
This is basic rule-based sentiment analysis. It classifies text as 'Positive', 'Negative', or 'Neutral' by counting predefined positive and negative words. The sentiment is determined by comparing the counts of positive versus negative keywords found in the text. It also measures the execution time.

### Use Case:
Rule-based sentiment analysis is a straightforward approach often used for initial sentiment detection, especially in domains where a specific set of keywords strongly correlates with sentiment (e.g., product reviews with terms like 'excellent' or 'terrible'). While less sophisticated than machine learning models, it's fast and interpretable, making it suitable for simple, quick assessments or as a baseline.

```python
import time

positive_words = {"happy", "good", "great", "excellent", "love", "amazing"}
negative_words = {"sad", "bad", "terrible", "hate", "awful", "worst"}

def sentiment_analysis(text):
    start_time = time.time()
    words = text.lower().split()
    pos_count = sum(1 for word in words if word in positive_words)
    neg_count = sum(1 for word in words if word in negative_words)

    sentiment = "Positive Sentiment" if pos_count > neg_count else "Negative Sentiment" if neg_count > pos_count else "Neutral Sentiment"

    end_time = time.time()
    execution_time = end_time - start_time

    return sentiment, execution_time

text = """
I'm happy that it had rain yesterday, but feeling sad because I had my football match
"""
sentiment, exec_time = sentiment_analysis(text)
print("Sentiment:", sentiment)
print("Execution Time: {:.6f} seconds".format(exec_time))
```
